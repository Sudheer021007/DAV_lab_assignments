Earthquke Data Analysis-Graph Explanation


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns # Import seaborn for color palettes

# Load the dataset
df = pd.read_csv("/content/earthquake_1995-2023.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Check columns
print(df.columns)

# Count each magnitude type
magtype_counts = df["magType"].value_counts()

# Print the counts
print(magtype_counts)

# Create bar graph
plt.figure(figsize=(10, 6)) # Increased figure size for better readability

# Get a color palette with enough colors for each magnitude type
colors = sns.color_palette('viridis', n_colors=len(magtype_counts))

plt.bar(
    magtype_counts.index,
    magtype_counts.values,
    color=colors # Apply the color palette
)

plt.title("Earthquakes by Magnitude Type", fontsize=14, fontweight='bold') # Added font styling
plt.xlabel("Magnitude Type", fontsize=12)
plt.ylabel("Number of Earthquakes", fontsize=12)

plt.xticks(rotation=45, ha='right') # Rotate and align x-axis labels
plt.grid(axis='y', linestyle='--', alpha=0.7) # Add a subtle y-grid
plt.tight_layout() # Adjust layout to prevent labels from being cut off

plt.show()
```

### Explanation of 'Earthquakes by Magnitude Type' Bar Chart

*   This chart shows how many earthquakes fall into different categories based on their 'Magnitude Type'.
*   Each bar represents a specific type of earthquake measurement, such as 'mww' or 'mwc'.
*   The height of the bar indicates the total number of earthquakes recorded for that particular type.
*   It helps us quickly see which magnitude types are most common in our dataset.
*   For example, 'mww' is clearly the most frequently observed magnitude type in this data.

```
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(df["magnitude"], bins=10)

plt.title("Distribution of Earthquake Magnitude")
plt.xlabel("Magnitude")
plt.ylabel("Frequency")

plt.show()
```

```
import matplotlib.pyplot as plt

# Prepare data for the pie chart using magtype_counts
labels = magtype_counts.index
sizes = magtype_counts.values

# Create the pie chart
plt.figure(figsize=(10, 8))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140, pctdistance=0.85, wedgeprops={'edgecolor': 'black'})
plt.title('Proportion of Earthquake Magnitude Types')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

### Explanation of 'Proportion of Earthquake Magnitude Types' Pie Chart

*   This pie chart visually represents the percentage share of each earthquake magnitude type.
*   It shows the same data as the first bar chart but emphasizes the proportion of each category.
*   Each 'slice' of the pie corresponds to a different magnitude type, like 'mww' or 'mwc'.
*   The size of each slice is proportional to the number of earthquakes in that magnitude type.
*   It clearly illustrates that 'mww' is the dominant magnitude type, accounting for approximately 50.2% of all entries.

### Scatter Plot: Magnitude vs. Depth
Let's visualize the relationship between earthquake magnitude and depth. This can help us understand if deeper earthquakes tend to have different magnitudes.

```
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(x='magnitude', y='depth', data=df, hue='magType', size='magnitude', sizes=(20, 400), alpha=0.6)
plt.title('Earthquake Magnitude vs. Depth')
plt.xlabel('Magnitude')
plt.ylabel('Depth (km)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Explanation of 'Earthquake Magnitude vs. Depth' Scatter Plot

*   This scatter plot examines the relationship between an earthquake's strength ('Magnitude') and how deep it occurred ('Depth').
*   Every single dot on the plot represents one earthquake event.
*   The horizontal position of a dot shows its magnitude, and the vertical position shows its depth.
*   Different colors are used to distinguish between various 'magType' categories.
*   The size of each dot also corresponds to the earthquake's magnitude, making stronger earthquakes visually larger.

```
 Convert 'date_time' to datetime objects and extract the date
df['date_time'] = pd.to_datetime(df['date_time'])
df['date'] = df['date_time'].dt.date

# Group by date and count earthquakes
daily_counts = df.groupby('date').size().reset_index(name='count')
daily_counts['date'] = pd.to_datetime(daily_counts['date'])

plt.figure(figsize=(14, 7))
sns.lineplot(x='date', y='count', data=daily_counts)
plt.title('Daily Earthquake Counts (1995-2023)')
plt.xlabel('Date')
plt.ylabel('Number of Earthquakes')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()#
```

### Explanation of 'Daily Earthquake Counts (1995-2023)' Line Plot

*   This line plot illustrates the number of earthquakes that happened each day over a long period, from 1995 to 2023.
*   The horizontal axis ('Date') shows the progression of time, from earlier years on the left to later years on the right.
*   The vertical axis ('Number of Earthquakes') indicates how many earthquakes were recorded on any given day.
*   The line connects the daily counts, allowing us to see trends and fluctuations in earthquake activity over time.
*   This graph helps us identify periods with higher or lower earthquake frequencies and potential temporal patterns.
